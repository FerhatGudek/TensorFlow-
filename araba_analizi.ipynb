```python
y = dataFrame["price"].values
x = dataFrame.drop("price",axis=1).values
```

```python
y
```

```python
x
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3,random_state=10)
```

```python
len(x_train)
```

```python
len(x_test)
```

```python
len(y_train)
```

```python
len(y_test)
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler()
```

```python
x_train = scaler.fit_transform(x_train)
x_test = scaler.fit_transform(x_test)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
model = Sequential() # bu ml nin giriş evresini oluşturur
model.add(Dense(12,activation="relu"))
model.add(Dense(12,activation="relu")) # bunlar nöron bağlarıdır
model.add(Dense(12,activation="relu"))
model.add(Dense(1)) # çıkış evresi
model.compile(optimizer="adam",loss="mse")
```

```python
model.fit(x=x_train,y=y_train,validation_data=(x_test,y_test),batch_size=250,epochs=300)
```

```python
kayipVerisi = pd.DataFrame(model.history.history)
kayipVerisi
```

```python
kayipVerisi.plot() # grafik oldukça iyi sonuç verdiğini söylüyor
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error
```

```python
tahminDizisi = model.predict(x_test) # bu aşamada öğretilen ml ye göre
# x_test'i tahmin ettirdik ve bize sonuç verdi şimdi o sonucu karsltrcz
```

```python
tahminDizisi
```

```python
mean_absolute_error(y_test,tahminDizisi)
```
