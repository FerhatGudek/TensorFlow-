```python
y = dataFrame["Type"].values
x = dataFrame.drop("Type",axis=1).values
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3,random_state=15)
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler()
```

```python
scaler.fit(x_train)
```

```python
x_train = scaler.transform(x_train)
```

```python
x_test = scaler.transform(x_test)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Activation , Dropout
from tensorflow.keras.callbacks import EarlyStopping
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
model = Sequential()
model.add(Dense(units=30, activation="relu"))
model.add(Dense(units = 15, activation="relu"))
model.add(Dense(units =15, activation="relu"))
model.add(Dense(units = 1, activation="sigmoid"))
model.compile(loss="binary_crossentropy",optimizer="adam")
```

```python
model.fit(x=x_train, y=y_train, epochs=700,validation_data=(x_test,y_test),verbose=1)
```

```python
modelKaybi = pd.DataFrame(model.history.history)
modelKaybi.plot()
# burada loss ve val_loss u çok ayrı gördük overfitting söz konusu
```

```python
model = Sequential()
model.add(Dense(units=30,activation="relu"))
model.add(Dense(units=15, activation="relu"))
model.add(Dense(units=15, activation="relu"))
model.add(Dense(units=1,activation="sigmoid"))
model.compile(loss="binary_crossentropy",optimizer="adam")
```

```python
earlyStopping = EarlyStopping(monitor="val_loss",mode="min",verbose=1,patience=25)
```

```python
model.fit(x = x_train, y=y_train, epochs=700,validation_data=(x_test,y_test),verbose=1,callbacks=[earlyStopping])
```

```python
modelKaybi = pd.DataFrame(model.history.history)
modelKaybi.plot() #early stopping bozulmaya başladığında durdurur
# grafik birbirine hiç olmadı daha çok yaklaştı diyebiliriz
```

```python
model = Sequential()
model.add(Dense(units=30, activation="relu"))
model.add(Dropout(0.6))
model.add(Dense(units=15, activation="relu"))
model.add(Dropout(0.6))
model.add(Dense(units=15, activation="relu"))
model.add(Dropout(0.6))
model.add(Dense(units=1, activation="sigmoid"))
model.compile(loss="binary_crossentropy",optimizer="adam")
```

```python
model.fit(x=x_test, y=y_test, epochs=700,validation_data=(x_test,y_test),verbose=1,callbacks=[earlyStopping])
```

```python
kayipDf = pd.DataFrame(model.history.history)
kayipDf.plot()
```

```python
# ...existing code...
# model.predict_classes yerine:
tahmin_prob = model.predict(x_test)
tahminlerimiz = (tahmin_prob > 0.5).astype(int)
print(tahminlerimiz)
# ...existing code...
```

```python
from sklearn.metrics import classification_report, confusion_matrix
```

```python
print(classification_report(y_test,tahminlerimiz))
# cokta başarılı olmayan bir tahmin mekanızması oluşturmuşuz :)
```

```python
print(confusion_matrix(y_test,tahminlerimiz))
```
